# 9. У какого процента студентов (по школам) и мать и отец имеют высшее
#образование?

```python
high_ed = df[(df["Medu"] == 4) & (df["Fedu"] == 4)]
(high_ed.groupby("school").size() / df.groupby("school").size()) * 100
```

# 10. У какого процента студентов (по школам) и мать и отец вообще не
#имеют образования?

```python
no_ed = df[(df["Medu"] == 0) & (df["Fedu"] == 0)]
no_ed.groupby("school").size().div(df.groupby("school").size(), fill_value=0) * 100
```

# 11. Какой процент родителей, имеющих высшее образование, проживает
#совместно?

```python
parents_high = df[(df["Medu"] == 4) | (df["Fedu"] == 4)]
(parents_high["Pstatus"].value_counts(normalize=True) * 100)["T"]
```

# 12. Определите процент семей, в которых один из родителей имеет высшее
#образование, а другой не имеет, но проживают совместно

```python
mixed = df[((df["Medu"] == 4) & (df["Fedu"] != 4)) | ((df["Medu"] != 4) & (df["Fedu"] == 4))]
mixed_together = mixed[mixed["Pstatus"] == "T"]
len(mixed_together) / len(df) * 100
```

# 13. Какое качество семейных отношений в этих семьях?
