Проанализируем датасет `student-mat`. Датасет содержит сведения о студентах двух школ. Смысл задачи — определять, основываясь на этих данных, итоговую оценку студента. Набор данных включает следующие атрибуты:


- **school** – школа ('GP' – Gabriel Pereira или 'MS' – Mousinho da Silveira).  
- **sex** – пол студента ('F' – female или 'M' – male).  
- **age** – возраст студента.  
- **address** – домашний адрес студента ('U' – городской или 'R' – сельский).  
- **famsize** – размер семьи ('LE3' – меньше или равно 3 или 'GT3' – больше чем 3).  
- **Pstatus** – статус совместного проживания родителей ('T' – совместное проживание или 'A' – раздельное проживание).  
- **Medu** – образование матери (0 – нет, 1 – начальное (4 класс), 2 – 9 классы, 3 – среднее или 4 – высшее).  
- **Fedu** – образование отца (0 – нет, 1 – начальное (4 класс), 2 – 9 классы, 3 – среднее или 4 – высшее).  
- **Mjob** – работа матери (учитель, медицинское обслуживание, связанные с гражданскими услугами (административными или полицейскими), на дому или другое).  
- **Fjob** – работа отца (учитель, медицинское обслуживание, связанные с гражданскими услугами (административными или полицейскими), на дому или другое).  
- **famrel** – качество семейных отношений (от 1 – очень плохо до 5 – отлично).  
- **freetime** – свободное время после школы (от 1 – очень мало до 5 – очень много).  
- **goout** – встречи с друзьями (от 1 – очень мало до 5 – очень много).  
- **Dalc** – потребление алкоголя в рабочий день (Daily Alcohol Consumption) (от 1 – очень низкое до 5 – очень высокое).  
- **Walc** – еженедельное потребление алкоголя (Weekly Alcohol Consumption) (от 1 – очень низкое до 5 – очень высокое).  
- **health** – текущее состояние здоровья (от 1 – очень плохое до 5 – очень хорошее).  
- **absences** – количество пропусков занятий в школе (от 0 до 93).  
- **G1** – оценка за первый период (от 0 до 20).  
- **G2** – оценка за второй период (от 0 до 20).  
- **G3** – итоговая оценка (от 0 до 20, целевой результат).

## 9. У какого процента студентов (по школам) и мать и отец имеют высшее
##образование?

```python
high_ed = df[(df["Medu"] == 4) & (df["Fedu"] == 4)]
(high_ed.groupby("school").size() / df.groupby("school").size()) * 100
```

## 10. У какого процента студентов (по школам) и мать и отец вообще не
##имеют образования?

```python
no_ed = df[(df["Medu"] == 0) & (df["Fedu"] == 0)]
no_ed.groupby("school").size().div(df.groupby("school").size(), fill_value=0) * 100
```

В данном датасете нет студентов, у которых мать и отец вообще не имеют образования.

## 11. Какой процент родителей, имеющих высшее образование, проживает
##совместно?

```python
parents_high = df[(df["Medu"] == 4) | (df["Fedu"] == 4)]
print("Процент таких семей",float((parents_high["Pstatus"].value_counts(normalize=True) * 100)["T"]))
```

## 12. Определите процент семей, в которых один из родителей имеет высшее
##образование, а другой не имеет, но проживают совместно

```python
mixed = df[((df["Medu"] == 4) & (df["Fedu"] != 4)) | ((df["Medu"] != 4) & (df["Fedu"] == 4))]
mixed_together = mixed[mixed["Pstatus"] == "T"]
print("Процент таких семей: ",len(mixed_together) / len(df) * 100)
```

## 13. Какое качество семейных отношений в этих семьях?

```python
mixed_together["famrel"].describe()
```

В среднем качество семейных отношений в таких семьях в районе 4.
